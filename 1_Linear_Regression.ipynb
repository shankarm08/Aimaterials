# Step 1 : Importing the libraries

```python
import numpy as np
import pandas as pd.
```

# Step 2 : Reading the dataset

```python
dataset = pd.read_csv("housing_prices_SLR.csv") # first we need upload the dataset
dataset.head() # will shoe first 5 rows
```

# Step 3 : Seperating the features and target

```python
x = dataset.iloc[:,0:1]
y = dataset.iloc[:,1]
x.head()
```

```python
y.head()
```

# Step 4 : Splitting the dataset into training set and test

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,train_size=0.8,random_state=42)
```

# Step 5 : Creating the Model

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

# Step 6 : Training Model 

```python
model.fit(x_train,y_train)
```

# Step 7 : Testing the Model

```python
samplepredciton = model.predict([[1240]])
print(samplepredciton)
```

```python
y_pred = model.predict(x_test) # predicting for all x_test
y_pred
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Hypothetical housing dataset
areas = np.array([1000, 1030, 1060, 1090])  # House areas in square feet
prices = np.array([5618, 5201, 4779, 5245])  # House prices in dollars

# Example of a simple linear regression model (hypothetical coefficients)
m = 200  # slope
b = 50000  # intercept

# Create a scatter plot of the data
plt.scatter(areas, prices, label='Actual Prices')

# Create a line plot of the linear regression model
x = np.linspace(min(areas), max(areas), 100)
y = m*x + b
plt.plot(x, y, label='Linear Regression Model', linestyle='--')

# Plot datapoints above the slope
above_slope = prices > m*areas + b
plt.scatter(areas[above_slope], prices[above_slope], color='red', label='Above Slope')

# Plot datapoints below the slope
below_slope = prices < m*areas + b
plt.scatter(areas[below_slope], prices[below_slope], color='blue', label='Below Slope')

# Add title and labels
plt.title('Linear Regression Model')
plt.xlabel('House Area (square feet)')
plt.ylabel('House Price (dollars)')

# Add legend
plt.legend()

# Display the plot
plt.show()
```
