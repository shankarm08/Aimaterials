# Step 1 : Importing the libraries

```python
import numpy as np
import pandas as pd
```

# Step 2 : Reading the dataset

```python
dataset = pd.read_csv("housing_prices_MLR.csv") # first we need upload the dataset
dataset.head() # will show first 5 rows
```

# Step 3 : Seperating the features and target

```python
x = dataset.iloc[:,0:3]
y = dataset.iloc[:,4]

#x = dataset[["AREA","FLOOR","ROOM"]]
#y = dataset["PRICE"]
x.head()
```

```python
y.head()
```

# Step 4 : Splitting the dataset into training set and test

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,train_size=0.8,random_state=42)
```

# Step 5 : Creating the Model

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

# Step 6 : Training Model 

```python
model.fit(x_train,y_train)
```

# Step 7 : Testing the Model

```python
samplepredciton = model.predict([[1000,7,1]])
print(samplepredciton)
```

```python
y_pred = model.predict(x_test)
y_pred
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Assuming your dataset is already loaded

# Example dataset
data = {
    'AREA': [1400, 1600, 1700, 1875, 1100],
    'FLOOR': [1, 2, 2, 1, 3],
    'ROOM': [3, 3, 2, 3, 1],
    'PRICE': [245000, 312000, 279000, 308000, 199000]
}

dataset = pd.DataFrame(data)

# Selecting features (X) and target variable (Y)
x = dataset.iloc[:, 0:3]
y = dataset.iloc[:, 3]

# Scatter plot
plt.figure(figsize=(12, 6))

plt.subplot(1, 3, 1)
plt.scatter(x['AREA'], y)
plt.title('Area vs Price')
plt.xlabel('Area')
plt.ylabel('Price')

plt.subplot(1, 3, 2)
plt.scatter(x['FLOOR'], y)
plt.title('Floor vs Price')
plt.xlabel('Floor')
plt.ylabel('Price')

plt.subplot(1, 3, 3)
plt.scatter(x['ROOM'], y)
plt.title('Room vs Price')
plt.xlabel('Room')
plt.ylabel('Price')

plt.tight_layout()
plt.show()
```
