```python
import numpy as np
import pandas as pd
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt
```

```python
dataset = pd.read_csv("Concrete_Data.csv") # first we need upload the dataset
dataset.head()
```

```python
x=dataset.iloc[:,0:8]
```

```python
x
```

```python
y=dataset.iloc[:,8:]
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,train_size=0.8,random_state=42)
```

```python
x.shape
```

```python
y.shape
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

```python
model.fit(x_train,y_train)
```

```python
samplepredciton = model.predict([[540,0,0,162,2.5,1040,676,28
]])
print(samplepredciton)
```

```python
y_pred = model.predict(x_test) # predicting for all x_test
y_pred
```

```python
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt
print("MSE:", mean_squared_error(y_test, y_pred))
print("R2 Score:", r2_score(y_test, y_pred))

# Plot
plt.scatter(y_test, y_pred, alpha=0.5)
plt.xlabel("Actual Strength")
plt.ylabel("Predicted Strength")
plt.title("Linear Regression: Concrete Strength")
plt.grid(True)
plt.show()
```

```python
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt
import numpy as np

print("MSE:", mean_squared_error(y_test, y_pred))
print("R2 Score:", r2_score(y_test, y_pred))

# Plot
plt.figure(figsize=(8, 6))
plt.scatter(y_test, y_pred, alpha=0.5, label="Predicted vs Actual")
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', label='Ideal Fit')  # Regression line
plt.xlabel("Actual MPG")
plt.ylabel("Predicted MPG")
plt.title("Linear Regression: Auto MPG")
plt.grid(True)
plt.legend()
plt.show()
```
